```python
import matplotlib.pyplot as plt
```

```python
with open("componentes.txt") as f:
    tamanhos = [int(linha) for linha in f if linha.strip()]
```

```python
x = list(range(1, len(tamanhos) + 1))

plt.figure(figsize=(8, 5))
plt.bar(x, tamanhos, color='steelblue')

for i, tam in zip(x, tamanhos):
    plt.text(i, tam + 5, str(tam), ha='center', fontsize=11)

plt.title('Distribuição dos Componentes Conexos')
plt.xlabel('Total de Componentes Encontrados')
plt.ylabel('Número de vértices')
plt.xticks(x, [f'Componente {i}' for i in x])
plt.yticks(range(0, max(tamanhos) + 100, 50))
plt.ylim(0, max(tamanhos) + 60)
plt.tight_layout()
plt.savefig('histograma.png', dpi=150)
plt.show()
```
