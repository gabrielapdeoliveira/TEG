```python
import matplotlib.pyplot as plt
```

```python
with open("componentes.txt") as f:
    tamanhos = [int(linha) for linha in f]
```

```python
plt.figure(figsize=(10, 6))
plt.hist(tamanhos, bins=20, color='steelblue', edgecolor='white')
plt.title('Distribuição dos Componentes Conexos')
plt.xlabel('Tamanho do componente (nº de vértices)')
plt.ylabel('Quantidade de componentes')
plt.tight_layout()
plt.savefig('histograma.png', dpi=150)
plt.show()
print(f"Total de componentes: {len(tamanhos)}")
print(f"Maior componente: {max(tamanhos)} vertices")
print(f"Menor componente: {min(tamanhos)} vertices")
```
